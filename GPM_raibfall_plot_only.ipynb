```python
import os
import glob
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# ======================================================
# Input & Output Folder
# ======================================================
input_folder = r"Y:\NERDRR\Team_NERDRR_2026\July 2026\Meghalaya flood\rainfall\Raifall Daily"

output_folder = os.path.join(input_folder, "Plots")
os.makedirs(output_folder, exist_ok=True)

# ======================================================
# Publication Style
# ======================================================
plt.rcParams.update({
    "font.family": "Times New Roman",
    "font.size": 13,
    "axes.linewidth": 1.5,
    "xtick.direction": "in",
    "ytick.direction": "in",
})

# ======================================================
# Read all CSV files
# ======================================================
files = glob.glob(os.path.join(input_folder, "*.csv"))

print(f"Found {len(files)} rainfall files.\n")

for file in files:

    print(f"Processing: {os.path.basename(file)}")

    # --------------------------------------------------
    # Read CSV
    # --------------------------------------------------
    df = pd.read_csv(file)

    # Remove extra spaces from column names
    df.columns = df.columns.str.strip()

    # Check required columns
    if "Date" not in df.columns or "Daily_Rainfall_mm" not in df.columns:
        print(f"Skipping {os.path.basename(file)} (required columns missing)")
        continue

    # --------------------------------------------------
    # Convert Date & Rainfall
    # --------------------------------------------------
    df["Date"] = pd.to_datetime(
        df["Date"].astype(str).str.strip(),
        format="mixed",
        dayfirst=True,
        errors="coerce"
    )

    df["Daily_Rainfall_mm"] = pd.to_numeric(
        df["Daily_Rainfall_mm"],
        errors="coerce"
    )

    df = df.dropna(subset=["Date", "Daily_Rainfall_mm"])

    df = df.sort_values("Date")

    # --------------------------------------------------
    # District Name
    # --------------------------------------------------
    district = os.path.splitext(os.path.basename(file))[0]
    district = district.replace("_DailyRainfall", "")

    # ==================================================
    # Create Figure
    # ==================================================
    fig, ax = plt.subplots(figsize=(12, 5))

    # Rainfall Area
    ax.fill_between(
        df["Date"],
        df["Daily_Rainfall_mm"],
        color="deepskyblue",
        alpha=0.25
    )

    # Rainfall Line
    ax.plot(
        df["Date"],
        df["Daily_Rainfall_mm"],
        color="#0055A4",
        linewidth=2.8,
        marker="o",
        markersize=6,
        markerfacecolor="white",
        markeredgewidth=1.5
    )

    # ==================================================
    # Axis Labels (Bold)
    # ==================================================
    ax.set_xlabel(
        "Date",
        fontsize=15,
        fontweight="bold"
    )

    ax.set_ylabel(
        "Rainfall (mm/day)",
        fontsize=15,
        fontweight="bold"
    )

    # ==================================================
    # Text inside graph
    # ==================================================
    ax.text(
        0.02,
        0.96,
        "IMERG Multi-Satellite Data",
        transform=ax.transAxes,
        fontsize=13,
        fontweight="bold",
        ha="left",
        va="top",
        bbox=dict(
            facecolor="white",
            edgecolor="black",
            alpha=0.8,
            boxstyle="round,pad=0.3"
        )
    )

    # ==================================================
    # Date Formatting
    # ==================================================
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%d-%b"))
    ax.xaxis.set_major_locator(mdates.AutoDateLocator())

    plt.xticks(rotation=45)

    # ==================================================
    # Grid
    # ==================================================
    ax.grid(
        True,
        linestyle="--",
        linewidth=0.6,
        alpha=0.4
    )

    # ==================================================
    # Full Border
    # ==================================================
    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_linewidth(1.5)
        spine.set_color("black")

    # ==================================================
    # Bold Tick Labels
    # ==================================================
    ax.tick_params(
        axis="both",
        labelsize=12,
        width=1.5,
        length=6
    )

    for tick in ax.get_xticklabels():
        tick.set_fontweight("bold")

    for tick in ax.get_yticklabels():
        tick.set_fontweight("bold")

    # ==================================================
    # Layout
    # ==================================================
    plt.tight_layout()

    # ==================================================
    # Save Figure
    # ==================================================
    outfile = os.path.join(
        output_folder,
        district + "_Rainfall.png"
    )

    plt.savefig(
        outfile,
        dpi=600,
        bbox_inches="tight",
        facecolor="white"
    )

    plt.close()

    print(f"Saved: {outfile}")

print("\n========================================")
print("All rainfall plots generated successfully.")
print(f"Output Folder: {output_folder}")
```

```python
import os
import glob
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# ======================================================
# Input and Output folders
# ======================================================
input_folder = r"Y:\NERDRR\Team_NERDRR_2026\July 2026\Meghalaya flood\rainfall\Raifall Daily"

output_folder = os.path.join(input_folder, "Plots")
os.makedirs(output_folder, exist_ok=True)

# ======================================================
# Publication Style
# ======================================================
plt.rcParams.update({
    "font.family": "Times New Roman",
    "font.size": 13,
    "axes.linewidth": 1.5,
    "xtick.direction": "out",
    "ytick.direction": "out"
})

# ======================================================
# Process all CSV files
# ======================================================
csv_files = glob.glob(os.path.join(input_folder, "*.csv"))

print(f"Found {len(csv_files)} files.\n")

for file in csv_files:

    # Read CSV
    df = pd.read_csv(file)

    # Clean column names
    df.columns = df.columns.str.strip()

    # Convert date
    df["Date"] = pd.to_datetime(
        df["Date"].astype(str).str.strip(),
        format="mixed",
        dayfirst=True,
        errors="coerce"
    )

    # Convert rainfall
    df["Daily_Rainfall_mm"] = pd.to_numeric(
        df["Daily_Rainfall_mm"],
        errors="coerce"
    )

    # Remove invalid rows
    df = df.dropna(subset=["Date", "Daily_Rainfall_mm"])

    # Sort by date
    df = df.sort_values("Date")

    # District name
    district = os.path.splitext(os.path.basename(file))[0]
    district = district.replace("_DailyRainfall", "")

    # ==================================================
    # Plot
    # ==================================================
    fig, ax = plt.subplots(figsize=(10, 4.5))

    ax.plot(
        df["Date"],
        df["Daily_Rainfall_mm"],
        color="red",
        linewidth=2.2
    )

    # ==================================================
    # Axis Labels (Bold)
    # ==================================================
    ax.set_xlabel(
        "Date",
        fontsize=14,
        fontweight="bold"
    )

    ax.set_ylabel(
        "Rainfall (mm/day)",
        fontsize=14,
        fontweight="bold"
    )

    # ==================================================
    # IMERG Label inside graph
    # ==================================================
    # ==================================================
    # IMERG Label inside graph (Top Right)
    # ==================================================
    ax.text(
        0.98,
        0.98,
        "IMERG Multi-Satellite Data",
        transform=ax.transAxes,
        fontsize=12,
        fontweight="bold",
        ha="right",
        va="top",
        bbox=dict(
            facecolor="white",
            edgecolor="black",
            boxstyle="round,pad=0.25",
            alpha=0.9
        )
    )

    # ==================================================
    # Date formatting
    # ==================================================
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%d-%b"))
    ax.xaxis.set_major_locator(mdates.AutoDateLocator())

    fig.autofmt_xdate(rotation=45)

    # ==================================================
    # Grid
    # ==================================================
    ax.grid(
        True,
        linestyle="--",
        linewidth=0.5,
        alpha=0.4
    )

    # ==================================================
    # Full Border
    # ==================================================
    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_linewidth(1.5)
        spine.set_color("black")

    # ==================================================
    # Bold Tick Labels
    # ==================================================
    ax.tick_params(
        axis='both',
        labelsize=11,
        width=1.3,
        length=5
    )

    for label in ax.get_xticklabels():
        label.set_fontweight("bold")

    for label in ax.get_yticklabels():
        label.set_fontweight("bold")

    # ==================================================
    # Layout
    # ==================================================
    plt.tight_layout()

    # ==================================================
    # Save figure
    # ==================================================
    outfile = os.path.join(
        output_folder,
        district + "_Rainfall_Line.png"
    )

    plt.savefig(
        outfile,
        dpi=600,
        bbox_inches="tight",
        facecolor="white"
    )

    plt.close()

    print(f"Saved: {outfile}")

print("\n===================================")
print("All rainfall plots generated successfully.")
print(f"Output Folder: {output_folder}")
```
