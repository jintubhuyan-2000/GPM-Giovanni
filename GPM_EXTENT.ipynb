```python
import geopandas as gpd
import pandas as pd

# =====================================
# USER INPUTS
# =====================================

# Path to your district shapefile
shapefile = r"D:\NER_DISTRICT_BOUNDARY_298\NER_DISTRICT_BOUNDARY_298_UTM_46.shp"

# Name of the district
district_name = "Kamrup Metro"

# Name of the district field in the shapefile
district_field = "DISTRICT"      # Change if your field name is different

# =====================================
# READ SHAPEFILE
# =====================================

gdf = gpd.read_file(shapefile)

print("Available Columns:")
print(gdf.columns)

# Convert to WGS84 if necessary
if gdf.crs != "EPSG:4326":
    gdf = gdf.to_crs(epsg=4326)

# =====================================
# SELECT DISTRICT
# =====================================

district = gdf[gdf[district_field].str.lower() == district_name.lower()]

if district.empty:
    print(f"District '{district_name}' not found.")
    exit()

# =====================================
# EXTRACT EXTENT
# =====================================

minx, miny, maxx, maxy = district.total_bounds

print("\n========== DISTRICT EXTENT ==========")
print("District :", district_name)
print(f"Min Longitude : {minx:.6f}")
print(f"Min Latitude  : {miny:.6f}")
print(f"Max Longitude : {maxx:.6f}")
print(f"Max Latitude  : {maxy:.6f}")

# =====================================
# SAVE CSV
# =====================================

extent = pd.DataFrame({
    "District":[district_name],
    "MinLon":[minx],
    "MinLat":[miny],
    "MaxLon":[maxx],
    "MaxLat":[maxy]
})

extent.to_csv("District_Extent.csv", index=False)

print("\nExtent saved as District_Extent.csv")
```
